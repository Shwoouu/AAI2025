```python
# ============================================
# STEP 1: PROMPT CHAINING
# Customer Support AI Simulation
# ============================================

customer_message = """
I ordered a pair of headphones two weeks ago, but they still
haven't arrived. The tracking number says the package has been
stuck in transit for five days. I need the headphones for a
trip next week.
"""

# STEP 1: Analyze the customer's problem
def analyze_issue(message):
    return {
        "problem": "The customer's headphones have not arrived.",
        "request": "The customer wants help locating the package.",
        "urgency": "High because the customer needs the headphones next week."
    }

issue = analyze_issue(customer_message)

print("===== STEP 1: ISSUE ANALYSIS =====")
print("Problem:", issue["problem"])
print("Request:", issue["request"])
print("Urgency:", issue["urgency"])


# STEP 2: Create a response based on the analysis
def create_response(issue):
    return f"""
Hello,

I'm sorry that your headphones have been delayed.

I understand that the package has been stuck in transit for
five days and that you need the headphones for your trip next
week. We recommend checking the latest tracking information
and contacting the shipping carrier if the package remains
stuck.

We apologize for the inconvenience and appreciate your patience.
"""

draft_response = create_response(issue)

print("\n===== STEP 2: DRAFT RESPONSE =====")
print(draft_response)


# STEP 3: Improve the response
def improve_response(response):
    return """
Hello,

I'm sorry your headphones have been delayed. I understand how
frustrating this is, especially since you need them for your
trip next week.

Please check the latest tracking information and contact the
shipping carrier if there are no updates. If the package
continues to be delayed, customer support can help determine
the next available solution.

Thank you for your patience.
"""

final_response = improve_response(draft_response)

print("\n===== STEP 3: FINAL RESPONSE =====")
print(final_response)


print("\n===== PROMPT CHAIN =====")
print("Customer Message")
print("       ↓")
print("Issue Analysis")
print("       ↓")
print("Draft Response")
print("       ↓")
print("Improved Response")
```
