```python
# ============================================
# STEP 3: SELF-REFLECTION
# Improving an AI Summary
# ============================================

text = """
Artificial intelligence is becoming increasingly important in
healthcare. AI systems can help doctors analyze medical images,
identify possible abnormalities, organize patient information,
and support clinical decision-making. Machine learning can find
patterns in large amounts of medical data. However, AI systems
should support healthcare professionals rather than completely
replace them. Privacy, accuracy, bias, and patient safety are
also important concerns when using AI in healthcare.
"""


# STEP 1: Initial summary
print("===== STEP 1: INITIAL SUMMARY =====")

initial_summary = """
Artificial intelligence is becoming useful in healthcare because
it can help doctors analyze medical images, organize information,
and support decision-making. Machine learning can identify
patterns in medical data. However, AI should support healthcare
professionals, while privacy, accuracy, bias, and patient safety
remain important concerns.
"""

print(initial_summary)


# STEP 2: Self-reflection
print("\n===== STEP 2: SELF-REFLECTION =====")

critique = """
The summary includes the main ideas and is mostly accurate.
It explains how AI can help healthcare professionals and mentions
important concerns. However, it could be slightly clearer about
the fact that AI is intended to support rather than completely
replace healthcare professionals.
"""

print(critique)


# STEP 3: Improved summary
print("\n===== STEP 3: IMPROVED SUMMARY =====")

improved_summary = """
Artificial intelligence is increasingly being used in healthcare
to help doctors analyze medical images, identify abnormalities,
organize patient information, and support clinical decisions.
Machine learning can identify useful patterns in large amounts
of medical data. AI is intended to support healthcare
professionals rather than completely replace them, while
privacy, accuracy, bias, and patient safety remain important
concerns.
"""

print(improved_summary)


# Show the improvement process
print("\n===== SELF-REFLECTION PROCESS =====")

print("""
Original Text
      ↓
Initial Summary
      ↓
AI Critique
      ↓
Improved Summary
""")
```
